## 2... Creating target variables

```python
target_df = raw_data.copy()
```

```python
target_df['RoR_Perc'] = 100*(target_df['SP500_Close']-target_df['SP500_Open'])/target_df['SP500_Open']
```

```python
#_, bins = pd.qcut(target_df['RoR_Perc'], q=3, labels=[0, 1, 2], retbins=True)
#print("Cutoff values:", bins)
```

```python
# Calculate the 5th and 95th percentiles
low_cutoff = target_df['RoR_Perc'].quantile(0.05)
high_cutoff = target_df['RoR_Perc'].quantile(0.95)

# Apply the custom cutoff values using pd.cut
target_df['RoR_Intensity'] = pd.cut(
    target_df['RoR_Perc'],
    bins=[-float('inf'), low_cutoff, high_cutoff, float('inf')],
    labels=[0, 1, 2],
    include_lowest=True
)

# Print the cutoff values
print("Cutoff values: Low =", low_cutoff, ", High =", high_cutoff)
```

```python
# Creating target variable as a rolling sum of the last 4 weeks

target_df['Rolling_RoR'] = target_df['RoR_Perc'].rolling(window=4).sum()

# Define the condition: if 'RoR_Percentage' is non-negative
condition = target_df['Rolling_RoR'] >= target_df['Rolling_RoR'].mean() + target_df['Rolling_RoR'].std()

# Create a column that marks upward trends (1 for True, 0 for False)
target_df['Target_Rolling'] = condition.astype(int)

target_df['Target_Rolling_shifted'] = target_df['Target_Rolling'].shift(-4)

target_df.head()
```

# 3... Changing variables to Percentage Change

```python
target_df.info()
```

```python
cols_to_change = target_df.iloc[:, list(range(0, 41))].columns

# Loop through each column in the list
for col in cols_to_change:
    # Calculate percentage change
    pct_change = target_df[col].pct_change() * 100

    # Replace NaN values (from the first row) and infinity values
    pct_change.replace([np.inf, -np.inf], np.nan, inplace=True)  # Replace inf with NaN

    # Fill NaN values with 0 (from both first row or inf replacements)
    target_df[col + '_pct_change'] = pct_change.fillna(0)
```

```python
final_df = target_df.copy()
final_df.drop(cols_to_change, axis=1)
```

# 4 Data set oversampling and unsampling

## 4.1 Oversample

```python
# Count values per unique value for 'RoR_Intensity'
ror_intensity_counts = final_df['RoR_Intensity'].value_counts()

ror_intensity_counts
```

```python
#Oversample values for Outliers RoR_Intensity on which colum of Ror value is 0 and 2

from sklearn.utils import resample

# Separate majority and minority classes
ror_0 = final_df[final_df['RoR_Intensity'] == 0]
ror_2 = final_df[final_df['RoR_Intensity'] == 2]
ror_others = final_df[~final_df['RoR_Intensity'].isin([0,2])]

# Upsample minority class
ror_0_Oversampled = resample(ror_0,
                          replace=True,  # sample with replacement
                          n_samples=ror_others.shape[0], # to match majority class
                          random_state=123) # reproducible results

ror_2_Oversampled = resample(ror_2,
                          replace=True,  # sample with replacement
                          n_samples=ror_others.shape[0], # to match majority class
                          random_state=123) # reproducible results

# Combine majority class with upsampled minority class
Oversampled_df = pd.concat([ror_others, ror_0_Oversampled, ror_2_Oversampled])

# Display new class counts
Oversampled_df['RoR_Intensity'].value_counts()
```

```python
Oversampled_df.to_csv('Oversampled_df.csv')
```

## 4.2 undersample

```python
# Undersample

# Downsample the majority classes to match the minority class
n_samples = ror_0.shape[0]  # Use the size of the smallest class as the target size

ror_others_downsampled = resample(ror_others,
                               replace=False,  # sample without replacement
                               n_samples=n_samples,  # to match minority class
                               random_state=123) # reproducible results


# Combine minority class with downsampled majority class
downsampled_df = pd.concat([ror_0, ror_2, ror_others_downsampled])

# Display new class counts
downsampled_df['RoR_Intensity'].value_counts()
```

```python
downsampled_df.to_csv('downsampled_df.csv')
```
